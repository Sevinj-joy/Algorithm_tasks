# Task 3

a = 0

Convert each variable to a boolean using bool()

Print the original value and the converted boolean value

Add a comment explaining why each value becomes True or False

```python
a = 0

b = 1

c = ""

d = "Python"

e = None

print(a, bool(a))
print(b, bool(b)) # 1 becomes True because any non-zero number is considered True.
print(c, bool(c))
print(d, bool(d)) # A non-empty string becomes True because it contains characters.
print(e, bool(e))
```
