## Task 10: Custom Data Generator

**Topic:** Generators, `*args`, `**kwargs`, and `return`

Write a function called `filter_numbers()` that accepts any number of positional arguments and returns a generator that produces only the numbers greater than a specified minimum value.

**Requirements:**
- Use `*args` to accept any number of numbers.
- Use `min_value` as a keyword-only parameter with a default value of `0`.
- Use `yield` to generate numbers greater than `min_value`.
- Return the generator object from `filter_numbers()`.
- Write a separate function called `display_results()` that uses `**kwargs` to display additional information.
- Test your functions with different minimum values and additional keyword arguments.

**Example:**
```python
numbers = filter_numbers(
    1, 5, 10, -3, 7,
    min_value=4
)

for number in numbers:
    print(number)

display_results(
    status="Completed",
    count=3,
    message="Filtering finished"
)
```

**Expected Output:**
```text
5
10
7
status: Completed
count: 3
message: Filtering finished
```

**Important:** Remember that `return` sends a result back to the caller, while `yield` produces values one at a time and allows a function to behave as a generator.
