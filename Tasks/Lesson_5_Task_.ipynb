# Task 3
Create two sets of animals: store_one = {'cat', 'dog', 'fish'} and store_two = {'fish', 'parrot', 'cat'}. Use a set operation to find and print the animals that are present in BOTH stores.


```python
store_one = {'cat', 'dog', 'fish'}
store_two = {'fish', 'parrot', 'cat'}

both_of = store_one.intersection(store_two)
print(both_of)
```

# Task 4
Using the same store_one and store_two sets, use a set operation to find and print the animals that are ONLY in store_one and not in store_two.



```python
store_one = {'cat', 'dog', 'fish'}
store_two = {'fish', 'parrot', 'cat'}

only_first_set = store_one.difference(store_two)
print(only_first_set)
```
