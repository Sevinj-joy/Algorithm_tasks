````python
import numpy as np


scores = np.array([
    [85, 92, 78, 90],
    [76, 88, 95, 82],
    [90, 91, 89, 94],
    [65, 72, 80, 75],
    [95, 98, 97, 96],
    [70, 75, 68, 80],
    [88, 84, 92, 89]
])

subjects = ["Math", "Physics", "Programming", "Statistics"]

telebe_ortalamasi = scores.mean(axis=1)
print("Student Average Scores:")
print(telebe_ortalamasi)
print(" -----------")
fenn_ortalamasi = scores.mean(axis=0)
print("Subject Average Scores:")
print(fenn_ortalamasi)
print(" -----------")

highest_average=np.max(telebe_ortalamasi)
print(highest_average)
print(" -----------")
lowest_average=np.min(telebe_ortalamasi)
print(lowest_average)
print(" -----------")

above_85=np.sum(telebe_ortalamasi>85)
print(f"Above 85 __ {above_85} student")
print(" -----------")

math_scores=scores[:,0] #mathscores
print(math_scores)
print(" -----------")


prog_scores=scores[:,2] #programingscores
print(prog_scores)
print(" -----------")

math_prog_score=scores[(math_scores>90) & (prog_scores>90)][:, [0, 2]]

print(math_prog_score)
print(" -----------")

at_least_80=np.sum(telebe_ortalamasi>=80)
print(f"At least 80 __ {at_least_80} student")
print(" -----------")

above_70=np.sum((scores>=70).all(axis=1))
print(f"Above 70 __ {above_70} student")
print(" -----------")
total_students=scores.shape[0]
print(f"Total students __ {total_students}")
print(" -----------")
percentage_of_total=above_70/total_students*100
print(f"Percentage of total __ {round(percentage_of_total)}%")
print(" -----------")


max_score = scores.max()
min_score = scores.min()
max_row, max_col=np.unravel_index(scores.argmax(), scores.shape)
min_row, min_col=np.unravel_index(scores.argmin(), scores.shape)

print(f"Ən yüksək bal: {max_score} -> Student {max_row + 1} , Subject: {subjects[max_col]}")
print(f"Ən aşağı bal:  {min_score} -> Student {min_row + 1} , Subject: {subjects[min_col]}")
````

````python
import numpy as np

temperatures = np.array([
    [20, 22, 25, 23, 21],
    [15, 17, 18, 16, 14],
    [30, 32, 35, 33, 31]
])

cities = ["City A", "City B", "City C"]

# F = C * 9/5 + 32

faranheight_temperatures=temperatures*9/5+32
print("Faranheight temperatures:")
print(faranheight_temperatures)
print("---------------")
correction = np.array([1.5, -2.0, 0.5]).reshape(-1,1)
corrected_temperatures = temperatures + correction
print("Corrected temperatures:")
print(corrected_temperatures)
print("---------------")

city_average_tempereatures=corrected_temperatures.mean(axis=1)
print("City average temperatures:")
print(city_average_tempereatures)
print("---------------")

day_average_temperatures=corrected_temperatures.mean(axis=0)
print("Day average temperatures:")
print(day_average_temperatures)
print("---------------")

above_30 = temperatures[temperatures > 30]
print("Above 30:")
print(above_30)
print("---------------")
below_18 = temperatures[temperatures > 18]
corrected_temperatures = np.maximum(corrected_temperatures, 18)
print("Below 18:")
print(below_18)
print("---------------")
bool_array=temperatures>temperatures.mean()
print("Boolean array:")
print(bool_array)
print("---------------")

# Explain:
# - What are the broadcasting rules?
# - Why can a `(3, 5)` array be combined with a `(3, 1)` array?
# - Why would combining a `(3, 5)` array with a `(4,)` array result in an error?
#NumPy compares array shapes from right to left using two rules:
#• Rule 1: If the arrays have different numbers of dimensions, the shape of the smaller array is padded with 1s on the left side.
#• Rule 2: Comparing each dimension from right to left, the sizes must either be equal or one of them must be 1. Otherwise, a ValueError is raised.
#____________________
# When comparing dimensions from right to left, both pairs are compatible:
# • Trailing dimensions (columns): 5 and 1 are compatible because one of them is 1 (NumPy stretches 1 to match 5).
# • Leading dimensions (rows): 3 and 3 are equal.
#____________________
# • First, by Rule 1, (4,) is padded on the left to become (1, 4).
# • Next, comparing from right to left, the trailing dimensions are 5 and 4.
# • Since 5 and 4 are neither equal nor is one of them 1, they are incompatible and NumPy raises a ValueError.
````
