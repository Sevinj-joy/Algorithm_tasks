````python
import numpy as np


scores = np.array([
    [85, 92, 78, 90],
    [76, 88, 95, 82],
    [90, 91, 89, 94],
    [65, 72, 80, 75],
    [95, 98, 97, 96],
    [70, 75, 68, 80],
    [88, 84, 92, 89]
])

subjects = ["Math", "Physics", "Programming", "Statistics"]

telebe_ortalamasi = scores.mean(axis=1)
print("Student Average Scores:")
print(telebe_ortalamasi)
print(" -----------")
fenn_ortalamasi = scores.mean(axis=0)
print("Subject Average Scores:")
print(fenn_ortalamasi)
print(" -----------")

highest_average=np.max(telebe_ortalamasi)
print(highest_average)
print(" -----------")
lowest_average=np.min(telebe_ortalamasi)
print(lowest_average)
print(" -----------")

above_85=np.sum(telebe_ortalamasi>85)
print(f"Above 85 __ {above_85} student")
print(" -----------")

math_scores=scores[:,0] #mathscores
print(math_scores)
print(" -----------")


prog_scores=scores[:,2] #programingscores
print(prog_scores)
print(" -----------")

math_prog_score=scores[(math_scores>90) & (prog_scores>90)][:, [0, 2]]

print(math_prog_score)
print(" -----------")

at_least_80=np.sum(telebe_ortalamasi>=80)
print(f"At least 80 __ {at_least_80} student")
print(" -----------")

above_70=np.sum((scores>=70).all(axis=1))
print(f"Above 70 __ {above_70} student")
print(" -----------")
total_students=scores.shape[0]
print(f"Total students __ {total_students}")
print(" -----------")
percentage_of_total=above_70/total_students*100
print(f"Percentage of total __ {round(percentage_of_total)}%")
print(" -----------")


max_score = scores.max()
min_score = scores.min()
max_row, max_col=np.unravel_index(scores.argmax(), scores.shape)
min_row, min_col=np.unravel_index(scores.argmin(), scores.shape)

print(f"Ən yüksək bal: {max_score} -> Student {max_row + 1} , Subject: {subjects[max_col]}")
print(f"Ən aşağı bal:  {min_score} -> Student {min_row + 1} , Subject: {subjects[min_col]}")
````

````python
import numpy as np

temperatures = np.array([
    [20, 22, 25, 23, 21],
    [15, 17, 18, 16, 14],
    [30, 32, 35, 33, 31]
])

cities = ["City A", "City B", "City C"]

# F = C * 9/5 + 32

faranheight_temperatures=temperatures*9/5+32
print("Faranheight temperatures:")
print(faranheight_temperatures)
print("---------------")
correction = np.array([1.5, -2.0, 0.5]).reshape(-1,1)
corrected_temperatures = temperatures + correction
print("Corrected temperatures:")
print(corrected_temperatures)
print("---------------")

city_average_tempereatures=corrected_temperatures.mean(axis=1)
print("City average temperatures:")
print(city_average_tempereatures)
print("---------------")

day_average_temperatures=corrected_temperatures.mean(axis=0)
print("Day average temperatures:")
print(day_average_temperatures)
print("---------------")

above_30 = temperatures[temperatures > 30]
print("Above 30:")
print(above_30)
print("---------------")
below_18 = temperatures[temperatures > 18]
corrected_temperatures = np.maximum(corrected_temperatures, 18)
print("Below 18:")
print(below_18)
print("---------------")
bool_array=temperatures>temperatures.mean()
print("Boolean array:")
print(bool_array)
print("---------------")

# Explain:
# - What are the broadcasting rules?
# - Why can a `(3, 5)` array be combined with a `(3, 1)` array?
# - Why would combining a `(3, 5)` array with a `(4,)` array result in an error?
#NumPy compares array shapes from right to left using two rules:
#• Rule 1: If the arrays have different numbers of dimensions, the shape of the smaller array is padded with 1s on the left side.
#• Rule 2: Comparing each dimension from right to left, the sizes must either be equal or one of them must be 1. Otherwise, a ValueError is raised.
#____________________
# When comparing dimensions from right to left, both pairs are compatible:
# • Trailing dimensions (columns): 5 and 1 are compatible because one of them is 1 (NumPy stretches 1 to match 5).
# • Leading dimensions (rows): 3 and 3 are equal.
#____________________
# • First, by Rule 1, (4,) is padded on the left to become (1, 4).
# • Next, comparing from right to left, the trailing dimensions are 5 and 4.
# • Since 5 and 4 are neither equal nor is one of them 1, they are incompatible and NumPy raises a ValueError.
````

````python
## Task 9 – Data Cleaning Using NumPy

# **Difficulty: Hard**

# A sensor system collects measurements, but some values are invalid due to hardware problems.

import numpy as np

sensor_data = np.array([
    [23.5, 24.1, -999, 25.3],
    [22.8, -999, 23.9, 24.5],
    [26.1, 25.7, 26.4, -999],
    [-999, 24.9, 25.2, 26.0],
    [23.1, 23.8, 24.2, 24.7]
])

INVALID_VALUE = -999
invalied_values=np.where(sensor_data==INVALID_VALUE)
print(invalied_values)
print("-------------------")

bool_array=sensor_data==INVALID_VALUE
print(bool_array)
print("-------------------")

total_invalid=np.sum(bool_array)
print(total_invalid)
print("-------------------")

percentage_invalid=(total_invalid/np.size(sensor_data))*100
print(percentage_invalid)
print("-------------------")

average_data=np.mean(sensor_data)
print(average_data)
print("-------------------")

replace_data=np.maximum(sensor_data,average_data)
print(replace_data)
print("-------------------")


calculate_average=np.mean(replace_data,axis=1)
print(calculate_average)
print("-------------------")
highest_score=np.argmax(calculate_average)
print(highest_score)
print("-------------------")
bool_mask=calculate_average>25
print(bool_mask)
print("-------------------")
valid_data=sensor_data[~bool_mask]
print(valid_data)
print("-------------------")

new_array=np.column_stack((calculate_average,sensor_data))
print(new_array)
print("-------------------")

#**Challenge:** Replace each invalid measurement with the average of its corresponding row instead of the global average.

# **Research challenge:** Investigate `np.nan`, `np.isnan()`, `np.nanmean()` and `np.nan_to_num()`.

# Explain the difference between representing missing values using a special number such as `-999` and using `np.nan`.

# • np.nan (Not a Number): Riyazi olaraq təyin olunmayan (məsələn, 0/0) və ya cədvəldə boş qalmış xanaları göstərən xüsusi float dəyəridir. Adi np.mean() kimi funksiyalar massivdə bircə dənə np.nan gördükdə yekun nəticəni də nan edir.
# • np.isnan(): Massivin hər bir elementini yoxlayır və np.nan olan yerlər üçün True, normal ədədlər üçün False qaytaran məntiqi (boolean) massiv yaradır. Qeyd edək ki, x == np.nan müqayisəsi həmişə False verdiyi üçün boş dəyərləri yalnız np.isnan() ilə tapmaq olar.
# • np.nanmean(): Massivdəki np.nan dəyərlərini görməzdən gələrək (saymadan) yalnız mövcud ədədlərin ədədi ortasını hesablayır.
# • np.nan_to_num(): Massivdəki bütün np.nan dəyərlərini standart olaraq 0.0 ilə (və ya nan=... parametri ilə verdiyiniz başqa bir ədədlə) əvəz edərək massivi təmizləyir.




# sensor_data[bool_array]=average_data
# print(sensor_data)
# print("-------------------")
# The value `-999` represents a missing or invalid measurement.

# **Your tasks:**

# 1. Identify all invalid measurements using Boolean indexing.
# 2. Count the total number of invalid measurements.
# 3. Calculate the percentage of missing measurements.
# 4. Replace all invalid measurements with the average of valid measurements in the entire dataset.
# 5. Calculate the average measurement for each sensor after cleaning.
# 6. Find the sensor with the highest average measurement.
# 7. Create a Boolean mask identifying all measurements greater than 25.
# 8. Create a new array containing only valid measurements.

# **Challenge:** Replace each invalid measurement with the average of its corresponding row instead of the global average.

# **Research challenge:** Investigate `np.nan`, `np.isnan()`, `np.nanmean()` and `np.nan_to_num()`.

# Explain the difference between representing missing values using a special number such as `-999` and using `np.nan`.

````
