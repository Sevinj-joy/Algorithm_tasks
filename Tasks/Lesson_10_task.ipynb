```python
# Task 2: Text Analyzer
# Given the following string:

# text = "  Python Programming Is Fun And Powerful  "
# Write a program that:

# Removes leading and trailing whitespace.
# Converts the text to lowercase.
# Counts the number of occurrences of the letter "p".
# Counts the number of words.
# Replaces "fun" with "interesting".
# Prints the cleaned string in reverse order.
# Uses a loop to count the vowels in the cleaned text.

text = "  Python Programming Is Fun And Powerful  "

removed_spaces=text.strip()
print(removed_spaces)
print(" ")
lower_case=removed_spaces.lower()
print(lower_case)
print(" ")
count_p=lower_case.count("p")
print(f"count of p is: {count_p}")
print(" ")

length_of_string=len(lower_case)
print(f"Length of text is: {length_of_string}")

replace_word=lower_case.replace("interesting","fun")
print(replace_word)
print(" ")

cleaned_text=replace_word[::-1]
print(cleaned_text)

for vowel in text:
  print(vowel)



```

```python
# Task 3: Product Code Parser

# You are given the following product code:

# product_code = "PRD-2025-047-89.99"
# Write a program that:

# Splits the code using "-" as the separator.
# Extracts the year, product ID, and price.
# Converts the year and product ID to integers and the price to a float.
# Prints each extracted value and its data type.
# Uses an if-else statement to determine whether the product price is greater than 50.

product_code = "PRD-2025-047-89.99"

split_code=product_code.split("-")
print(split_code,type(split_code))
print(" ")

year=int(split_code[1])
print(f"the year is: {year}")
print(" ")

id_product=split_code[2]
print(f"the ID is: {id_product}")
print(" ")

price=float(split_code[3])
print(f"the price is: {price}")
print(" ")

print(" ______The type of list elements______")
print(" ")
print(" The type of Year is: " ,type(year))
print(" The type of ID is: " ,type(id_product))
print(" The type of PRICE is: " ,type(price))
print(" ")

if price>50:
  print("The product price is greater than 50")

else:
  print("The product price is not greater than 50")
```

```python
#Task 6: Compare Course Enrolments

# Two groups of students enrolled in different courses:

# python_students = {"Alice", "Bob", "Charlie", "Diana", "Ethan"}
# ml_students = {"Charlie", "Diana", "Frank", "Grace", "Alice"}
# Write a program that finds:

# Students enrolled in both courses.
# Students enrolled only in Python.
# Students enrolled in at least one course.
# Students enrolled in exactly one of the two courses.
# The number of students in each resulting group.
# Print each result as a set.


python_students = {"Alice", "Bob", "Charlie", "Diana", "Ethan"}
ml_students = {"Charlie", "Diana", "Frank", "Grace", "Alice"}

print(" ")
both_courses=python_students.intersection(ml_students)
print(f"Students enrolled in both courses: {both_courses}")
print(" ")
only_python=python_students.difference(ml_students)
print(f"Students enrolled only in Python: {only_python}")
print(" ")
at_least=python_students.union(ml_students)
print(f"Students enrolled in at least one course: {at_least}")
print(" ")

unique_students=python_students.symmetric_difference(ml_students)
print(f"Students enrolled in exactly one of the two courses: {unique_students}")
print(" ")

count_python=len(python_students)
print(f"The number of students enrolled in Python: {count_python}")
print(" ")
count_ml=len(ml_students)
print(f"The number of students enrolled in ML: {count_ml}")

print(" ")
set_python=set(python_students)
set_ml=set(ml_students)
print(" ")
print(set_python)
print(" ")
print(set_ml)


```
