```python
#Task 6: Compare Course Enrolments

# Two groups of students enrolled in different courses:

# python_students = {"Alice", "Bob", "Charlie", "Diana", "Ethan"}
# ml_students = {"Charlie", "Diana", "Frank", "Grace", "Alice"}
# Write a program that finds:

# Students enrolled in both courses.
# Students enrolled only in Python.
# Students enrolled in at least one course.
# Students enrolled in exactly one of the two courses.
# The number of students in each resulting group.
# Print each result as a set.


python_students = {"Alice", "Bob", "Charlie", "Diana", "Ethan"}
ml_students = {"Charlie", "Diana", "Frank", "Grace", "Alice"}

print(" ")
both_courses=python_students.intersection(ml_students)
print(f"Students enrolled in both courses: {both_courses}")
print(" ")
only_python=python_students.difference(ml_students)
print(f"Students enrolled only in Python: {only_python}")
print(" ")
at_least=python_students.union(ml_students)
print(f"Students enrolled in at least one course: {at_least}")
print(" ")

unique_students=python_students.symmetric_difference(ml_students)
print(f"Students enrolled in exactly one of the two courses: {unique_students}")
print(" ")

count_python=len(python_students)
print(f"The number of students enrolled in Python: {count_python}")
print(" ")
count_ml=len(ml_students)
print(f"The number of students enrolled in ML: {count_ml}")

print(" ")
set_python=set(python_students)
set_ml=set(ml_students)
print(" ")
print(set_python)
print(" ")
print(set_ml)


```

```python
# Task 9: Student Performance Dictionary

# You are given the following dictionary:

# students = {
#     "Alice": [85, 90, 92],
#     "Bob": [70, 65, 80],
#     "Charlie": [95, 88, 91],
#     "Diana": [60, 75, 68]
# }
# Write a program that:

# Calculates the average score for each student.
# Stores the averages in a new dictionary.
# Uses if-else to classify each student as "Pass" if their average is at least 70, otherwise "Fail".
# Prints the name, average, and result of each student.
# Finds the student with the highest average score.


students = {
    "Alice": [85, 90, 92],
    "Bob": [70, 65, 80],
    "Charlie": [95, 88, 91],
    "Diana": [60, 75, 68]
}


student_averages = {
    name: sum(grades) / len(grades) for name, grades in students.items()
}

print(student_averages)
print(" ")

for key, value in student_averages.items():
  if value >= 70:
    print(f"Student: {key}, Average: {value}, Result: Pass")
  else:
    print(f"Student: {key}, Average: {value}, Result: Fail")

print(" ")

hightest_score=max(student_averages.values())
print(f"The highest average score is: {hightest_score}")
best_student=max(student_averages,key=student_averages.get)
print(f"The student is: {best_student}")
```
