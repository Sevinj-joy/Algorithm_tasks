```python
#Task 6: Compare Course Enrolments

# Two groups of students enrolled in different courses:

# python_students = {"Alice", "Bob", "Charlie", "Diana", "Ethan"}
# ml_students = {"Charlie", "Diana", "Frank", "Grace", "Alice"}
# Write a program that finds:

# Students enrolled in both courses.
# Students enrolled only in Python.
# Students enrolled in at least one course.
# Students enrolled in exactly one of the two courses.
# The number of students in each resulting group.
# Print each result as a set.


python_students = {"Alice", "Bob", "Charlie", "Diana", "Ethan"}
ml_students = {"Charlie", "Diana", "Frank", "Grace", "Alice"}

print(" ")
both_courses=python_students.intersection(ml_students)
print(f"Students enrolled in both courses: {both_courses}")
print(" ")
only_python=python_students.difference(ml_students)
print(f"Students enrolled only in Python: {only_python}")
print(" ")
at_least=python_students.union(ml_students)
print(f"Students enrolled in at least one course: {at_least}")
print(" ")

unique_students=python_students.symmetric_difference(ml_students)
print(f"Students enrolled in exactly one of the two courses: {unique_students}")
print(" ")

count_python=len(python_students)
print(f"The number of students enrolled in Python: {count_python}")
print(" ")
count_ml=len(ml_students)
print(f"The number of students enrolled in ML: {count_ml}")

print(" ")
set_python=set(python_students)
set_ml=set(ml_students)
print(" ")
print(set_python)
print(" ")
print(set_ml)


```

```python
# Task 9: Student Performance Dictionary

# You are given the following dictionary:

# students = {
#     "Alice": [85, 90, 92],
#     "Bob": [70, 65, 80],
#     "Charlie": [95, 88, 91],
#     "Diana": [60, 75, 68]
# }
# Write a program that:

# Calculates the average score for each student.
# Stores the averages in a new dictionary.
# Uses if-else to classify each student as "Pass" if their average is at least 70, otherwise "Fail".
# Prints the name, average, and result of each student.
# Finds the student with the highest average score.


students = {
    "Alice": [85, 90, 92],
    "Bob": [70, 65, 80],
    "Charlie": [95, 88, 91],
    "Diana": [60, 75, 68]
}


student_averages = {
    name: sum(grades) / len(grades) for name, grades in students.items()
}

print(student_averages)
print(" ")

for key, value in student_averages.items():
  if value >= 70:
    print(f"Student: {key}, Average: {value}, Result: Pass")
  else:
    print(f"Student: {key}, Average: {value}, Result: Fail")

print(" ")

hightest_score=max(student_averages.values())
print(f"The highest average score is: {hightest_score}")
best_student=max(student_averages,key=student_averages.get)
print(f"The student is: {best_student}")
```

```python

# Task 10: Word Frequency Counter

# Given the sentence:

# sentence = "python is easy and python is powerful and python is popular"

# Write a program that:

# Splits the sentence into individual words.
# Uses a dictionary to count how many times each word appears.
# Prints each word and its frequency.
# Identifies the most frequent word.
# Uses a loop rather than an external library.

sentence = "python is easy and python is powerful and python is popular"

words = sentence.split()

word_counts = {}
for word in words:
  if word in word_counts:
    word_counts[word] += 1
  else:
    word_counts[word] = 1

print("Word Frequencies:")
for word, count in word_counts.items():
  print(f"  - '{word}': {count}")

print("-" * 35)
```

```python
# Section 4: Comprehensions

# Task 11: Filter and Transform Numbers

# Given the list:

# numbers = [12, 7, 25, 30, 18, 9, 42, 15, 6, 33]
# Use list comprehensions to:

# Create a list of numbers divisible by 3.
# Create a list of squares of numbers greater than 15.
# Create a list containing "Even" for even numbers and "Odd" for odd numbers.
# Create a list of numbers that are both greater than 10 and divisible by 5.

numbers = [12, 7, 25, 30, 18, 9, 42, 15, 6, 33]
divisible_by_3=[i for i in numbers if i%3==0]
print(divisible_by_3)
print(" ")

squares_numbers=[i**2 for i in numbers if i>15]
print(squares_numbers)
print(" ")
even_odd_numbers=["Even" if i%2==0 else "Odd"for i in numbers]
print(even_odd_numbers)
print(" ")

greater_ten_and_five=[i for i in numbers if i>10 and i%5==0]
print(greater_ten_and_five)
```

```python
# Task 13: Nested List Comprehension

# Given the matrix:

# matrix = [
#     [1, 2, 3],
#     [4, 5, 6],
#     [7, 8, 9]
# ]
# Write a program that:

# Uses a nested list comprehension to flatten the matrix into a one-dimensional list.
# Creates a list containing only the even numbers from the matrix.
# Creates a new matrix in which every element is multiplied by 2.
# Calculates the sum of all elements in the original matrix.
matrix = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 9]
]
flatened_matrix=[item for row in matrix for item in row]
print(flatened_matrix)
print(" ")

even_numbers=[i for i in flatened_matrix if i%2==0]
print(even_numbers)
print(" ")

new_matrix=[[item*2 for item in row] for row in matrix]
print(new_matrix)
print(" ")

sum_of_elements=sum(flatened_matrix)
print(sum_of_elements)
```

```python
#Task 15: Multiplication Table

#Ask the user to enter an integer n.

#Write a program that uses a for loop to print the multiplication table of n from 1 to 12.

#Requirements:

#Display each multiplication in the format n x i = result.
#After printing the table, calculate and print the sum of all 12 products.
#Reject inputs that are not valid integers.

while True:
  try:
    n = int(input("Please enter a valid integer: "))
    break
  except ValueError:
    print("Invalid input! Please enter a valid integer.")

total_sum = 0


for i in range(1,13):
  result=n*i
  print(result)


total_sum += result
print(f"The sum of all 12 products is: {total_sum}")


```

```python
# Task 16: Guess the Number

# Create a number-guessing game using a while loop.

# Use the following secret number:

# secret_number = 37
# Requirements:

# Ask the user to guess the number.
# Print "Too low!" if the guess is smaller.
# Print "Too high!" if the guess is larger.
# Print a success message when the guess is correct.
# Count and display the total number of attempts.
# Handle invalid input without terminating the game.
secret_number = 37
total_attempts = 0

print("Welcome to the Number Guessing Game!")

while True:
  try:

    guess_number = int(input("Guess number: "))
    total_attempts += 1

    if guess_number < secret_number:
      print("Too low!")
    elif guess_number > secret_number:
      print("Too high!")
    else:
      print("Success!")
      break

  except ValueError:

    print("Invalid input! Please enter a valid integer.")

print(f"Total number of attempts: {total_attempts}")
```
