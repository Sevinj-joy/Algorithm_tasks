```python
# Task 1: Student Grade Calculator

# Ask the user to enter their name, exam score, and assignment score.

# Requirements:

# Convert the scores to floating-point numbers.
# Calculate the final grade using the following formula:
# Exam: 70% of the final grade.
# Assignment: 30% of the final grade.
# Print the student's name and final grade rounded to two decimal places.
# Use if-elif-else to assign a letter grade:
# A: 90 or above
# B: 80–89.99
# C: 70–79.99
# D: 60–69.99
# F: below 60
# If either score is outside the range 0–100, print an error message instead of calculating the grade.

name=input("Enter ur name please:")
exam_score=int(input("Enter ur exam score:"))
assignment_score=int(input("Enter ur assignment score:"))

if 0<=exam_score<=100 and 0<=assignment_score<=100:
    result = exam_score * 0.7 + assignment_score * 0.3
    print(f"Student: {name} | Final Grade: {result:.2f}")

    if result >=90:
     print("A")
    elif result>=80:
     print("B")
    elif result>=70:
     print("C")
    elif result>=60:
     print("D")
    elif result<60:
     print("F")

else:
 print("You cannot write out of range (0-100)")














```

```python
# Task 2: Text Analyzer
# Given the following string:

# text = "  Python Programming Is Fun And Powerful  "
# Write a program that:

# Removes leading and trailing whitespace.
# Converts the text to lowercase.
# Counts the number of occurrences of the letter "p".
# Counts the number of words.
# Replaces "fun" with "interesting".
# Prints the cleaned string in reverse order.
# Uses a loop to count the vowels in the cleaned text.

text = "  Python Programming Is Fun And Powerful  "

removed_spaces=text.strip()
print(removed_spaces)
print(" ")
lower_case=removed_spaces.lower()
print(lower_case)
print(" ")
count_p=lower_case.count("p")
print(f"count of p is: {count_p}")
print(" ")

length_of_string=len(lower_case)
print(f"Length of text is: {length_of_string}")

replace_word=lower_case.replace("interesting","fun")
print(replace_word)
print(" ")

cleaned_text=replace_word[::-1]
print(cleaned_text)

for vowel in text:
  print(vowel)



```
