```python
#task1
try:
    number1 = int(input("enter a number:"))
    number2 = int(input("enter a number:"))
    operator = input("Enter an operator (+, -, *, /): ").strip()

    if operator == "+":
        result = number1 + number2
        print(result)
    elif operator == "-":
        result = number1 - number2
        print(result)
    elif operator == "*":
        result = number1 * number2
        print(result)
    elif operator == "/":
        result = number1 / number2
        print(result)
    else:
        print("Error: Invalid operator. Please use +, -, *, or /.")

except ValueError:
    print("Error: Invalid input! Please enter valid numeric values.")

except ZeroDivisionError:
    print("Error: Division by zero is not allowed.")
```

```python
#TASK2
total_score = 0
number_students = 5

print(f"Please enter the scores for {number_students} students.")

for i in range(1, number_students + 1):
    while True:
        try:

            score = int(input(f"Enter score for Student {i} (0-100): "))


            if 0 <= score <= 100:
                total_score += score
                break
            else:
                print("Invalid score. Please enter a score between 0 and 100.")


        except ValueError:
            print("Error: Invalid input! Please enter a valid whole number.")

average_score = total_score / number_students

print("\n--- Final Results ---")
print(f"Total Combined Score: {total_score}")
print(f"Average Class Score:  {average_score:.2f}")

```

```python
#TASK3

numbers = [12, 25, 37, 48, 59]

try:

    new_index = int(input("Enter an index: "))
    element = numbers[new_index]

except IndexError:
    print("Error: Index out of range.")
except ValueError:
    print("Error: Please enter a valid integer for the index.")
else:

    print(f"Element: {element}")
    print(f"Square: {element ** 2}")
    print("Access completed successfully.")
finally:
    print(" ")
    print("List access attempt finished.")
```

```python
#task4

usernames = ["sibel", "aydan", "murad", "leyla", "kamran"]

try:

    username_example = input("Enter an username: ")
    position = usernames.index(username_example)
    print(f"Username found at position {position}.")

except IndexError:
    print("Error: Index out of range.")
except ValueError:
    print("Error: Username not found.")
finally:
    print("Search completed.")
```
