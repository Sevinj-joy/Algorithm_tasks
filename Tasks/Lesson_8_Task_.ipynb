## Task 9: Generate All Pairs with a Condition

Given a list of integers, generate all ordered pairs `(a, b)` where `a` and `b` are different numbers and their sum is equal to 10.


**Expected output:**
```python

```

**Challenge:** Modify your solution to include only pairs where `a < b`.

````python
#task1

square_list=[x**2 for x in range(1,21)]

print(square_list)
````

````python
#task2

numbers = [4, 8, 12, 15, 20, 9, 22, 10, 30, 7]

even_number_list=[x for x in numbers if x%2==0 and x>10]

print(even_number_list)
````

````python
#task3

words = ["python", "AI", "machine", "code", "data", "learning", "loop"]

new_list=[x.upper() for x in words if len(x)>4]
print(new_list)
````

````python
#task4
numbers=range(1,101)

new_list=[x for x in numbers if x%3==0 or x%5==0]

print(new_list)
````

````python
#task5

x_values = [1, 2, 3]
y_values = [10, 20, 30]

new_list=[(x, y) for x in x_values for y in y_values]
print(new_list)

````

````python
#task6

multiplication_table=[[x * y for y in range(1, 6)] for x in range(1,6)]

for row in multiplication_table:
    print(row)
````

````python
#task7

matrix = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 9]
]


flat_list = [item for sublist in matrix for item in sublist]

print(flat_list)
````
