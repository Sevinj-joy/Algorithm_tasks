# Task 9: Content Validation
Instructions: Test and print the validation checks to see if a mixed alphanumeric string contains strictly alphabetical letters, and whether a separate numeric string consists entirely of digits.

```python

str1 = "Python3"

str2 = "12345"

contains_only_letters = str1.isalpha()
contains_only_digits = str2.isdigit()

print(f"Is '{str1}' strictly alphabetical?_____ {contains_only_letters}")
print(f"Does '{str2}' consist entirely of digits?_____ {contains_only_digits}")
```

# Task 10: Advanced Slicing & Reversal
Instructions: Given a multi-word string, write code to extract characters at every alternate step position, and then reverse the entire sequence using a negative step parameter.

```python
text = "Python Programming"

alternate_chars = text[::2]

reversed_sequence = alternate_chars[::-1]

combined_result = text[::-2]

print(f"Original Text: '{text}'")
print(f"Extracted nd Reversed: '{reversed_sequence}'")
print(f"Combined single-step method:    '{combined_result}'")
```
